# Problem Statement

## The Problem

Modern fitness and wellness technology suffers from three interconnected limitations that this project aims to address:

**1. Fragmented Data Landscape**
Health and fitness data is scattered across siloed platforms — wearables track activity, MyFitnessPal tracks nutrition, sleep apps track recovery — yet no platform integrates these multi-modal signals into a unified analytical framework capable of supporting evidence-based training decisions.

**2. Absence of Scientific Rigour in Commercial Fitness Apps**
Most commercial fitness recommendation engines operate on rule-based heuristics ("eat 1g protein per pound of bodyweight") rather than statistically grounded, individualised predictions. The biological mechanism linking nutrition to muscle hypertrophy — specifically the mTOR-mediated protein synthesis pathway — is rarely modelled formally in software systems.

**3. Disregard for Historical Training Knowledge**
Bodybuilding has accumulated nearly a century of empirical training wisdom — from Eugen Sandow's symmetry principles (1890s) through Mike Mentzer's Heavy Duty HIT (1970s) to Chris Bumstead's modern classic physique (2020s). This historical knowledge remains locked in books, forums, and oral tradition rather than being computationally integrated into modern recommendation systems.

## The Proposed Solution

This project develops **HELIX** — an end-to-end intelligent platform that:

- **Integrates** heterogeneous wellness datasets (body composition, training behaviour, nutrition with amino acid profiles, sleep & recovery indicators) into a unified analytical framework
- **Predicts** muscle hypertrophy outcomes and recovery capacity using classical ML and deep learning models
- **Encodes** the protein synthesis biological pathway as a formal OWL knowledge graph queryable via SPARQL
- **Incorporates** structured historical training principles from four distinct bodybuilding eras as a knowledge layer informing recommendations
- **Deploys** the resulting system as a cloud-native microservice architecture on Google Cloud Platform

## Significance

The successful completion of this project will demonstrate the viability of combining classical statistical analysis, machine learning, semantic web technologies, and historical domain knowledge into a single coherent system for personalised athletic development — addressing a gap currently present in both academic literature and commercial fitness technology.

---# Section 3: Data Quality Assessment## 3.1 Quality DiagnosticsBefore cleaning, we systematically diagnose each dataset for:- **Missing values** — incomplete records that need handling- **Duplicate rows** — repeated entries that bias statistics- **Data type issues** — strings stored as numbers, or vice versa- **Memory footprint** — to gauge processing requirements

````python
import pandas as pd

raw = '/content/drive/MyDrive/HELIX_Project/data/raw'

datasets = {
    'Body Performance':  f'{raw}/body_performance/bodyPerformance.csv',
    'Exercise DB':       f'{raw}/exercise_db/megaGymDataset.csv',
    'Gym Members':       f'{raw}/gym_members/gym_members_exercise_tracking.csv',
    'Nutrition':         f'{raw}/nutrition/nutrients_csvfile.csv',
    'Sleep Health':      f'{raw}/sleep_health/Sleep_health_and_lifestyle_dataset.csv',
}

for name, path in datasets.items():
    print(f"\n{'='*70}")
    print(f"🔍 {name.upper()} — QUALITY CHECK")
    print(f"{'='*70}")

    df = pd.read_csv(path)

    print(f"\n📏 Total rows: {df.shape[0]:,}")
    print(f"📊 Total columns: {df.shape[1]}")
    print(f"💾 Memory: {df.memory_usage(deep=True).sum() / 1024:.1f} KB")

    # Missing values
    missing = df.isnull().sum()
    missing = missing[missing > 0]
    if len(missing) > 0:
        print(f"\n⚠️  Missing values:")
        for col, count in missing.items():
            pct = (count / len(df)) * 100
            print(f"   • {col}: {count:,} missing ({pct:.1f}%)")
    else:
        print(f"\n✅ No missing values!")

    # Duplicates
    dups = df.duplicated().sum()
    print(f"\n🔄 Duplicate rows: {dups}")

    # Data types
    print(f"\n📋 Data types:")
    type_counts = df.dtypes.value_counts()
    for dtype, count in type_counts.items():
        print(f"   • {dtype}: {count} columns")

print(f"\n\n{'='*70}")
print("✅ Quality check complete!")
print(f"{'='*70}")
````

# Section 2: Data Quality Assessment & Cleaning

## 2.1 Quality Findings

Initial inspection revealed the following data quality issues that required intervention before the data could be used for modelling:

| Dataset | Issue Detected | Action Taken |
|---|---|---|
| Body Performance | 1 duplicate row | Removed |
| Exercise Database | 53% missing descriptions, 65% missing ratings | Dropped non-essential columns |
| Gym Members | None — already clean | Standardised column naming only |
| Nutrition (original) | Corrupted placeholder values in multiple rows | Replaced with more reliable source |
| Sleep Health | 58% NaN in Sleep Disorder column | Filled as "None" (meaning no disorder, not missing data) |

## 2.2 The Nutrition Dataset Decision

During quality validation, the originally selected nutrition dataset was found to contain placeholder integer sequences (e.g. 230, 231, 232...) in place of measured macronutrient values for several rows. This was identified through a sanity-check protocol comparing dataset values against published nutritional reference standards.

Rather than attempt patch-level fixes that could compromise data integrity, the dataset was replaced with a more comprehensive USDA-derived nutrition database. The replacement contains **8,789 foods across 76 nutritional attributes**, including individual amino acid profiles — a capability that significantly enhances downstream knowledge graph construction in Phase 5.

This decision exemplifies a key principle in applied data science: **data validation should precede analysis, and methodological transparency should accompany every data substitution.**



## 3.2 Apply Cleaning Pipeline

We now apply targeted cleaning to each of the five datasets. Each dataset receives cleaning tailored to its specific quality issues identified above:

| Dataset | Cleaning Applied |
|---|---|
| Body Performance | Drop 1 duplicate row; standardise column names (`body fat_%` → `body_fat_pct`) |
| Exercise Database | Drop unreliable columns (Rating: 65% missing, Description: 53% missing); fill missing Equipment with "Bodyweight" |
| Gym Members | Standardise column names; no other cleaning needed (already pristine) |
| Nutrition (initial) | Convert text-numbers to numeric; replace trace values ('t') with 0; drop rows with missing critical values |
| Sleep Health | Fill missing Sleep Disorder values with "None" (semantically: not missing, but no disorder present); split combined Blood Pressure column into Systolic_BP and Diastolic_BP |

> **Methodological note on nutrition data:** The cleaning applied to the initial nutrition dataset in this cell becomes *retrospectively redundant* — in Section 5 we will identify integrity issues in this dataset that motivate its complete replacement. We retain this initial cleaning step in the notebook because (a) it documents the standard cleaning pipeline applied uniformly across datasets, and (b) the diagnostic failure of this cleaning step is itself evidence of the dataset's unsuitability. This is a deliberate methodological choice, not an oversight.

````python
import pandas as pd
import numpy as np
import os

raw = '/content/drive/MyDrive/HELIX_Project/data/raw'
clean = '/content/drive/MyDrive/HELIX_Project/data/processed'
os.makedirs(clean, exist_ok=True)

print("🧹 Starting the big clean...\n")

# ───────────────────────────────────────────────────────────
# 1. BODY PERFORMANCE — drop the 1 duplicate
# ───────────────────────────────────────────────────────────
print("1️⃣  Cleaning Body Performance...")
df = pd.read_csv(f'{raw}/body_performance/bodyPerformance.csv')
before = df.shape[0]
df = df.drop_duplicates()
df.columns = df.columns.str.strip().str.replace(' ', '_').str.replace('%', 'pct')
df.to_csv(f'{clean}/body_performance_clean.csv', index=False)
print(f"   ✅ Removed {before - df.shape[0]} duplicate. Final: {df.shape}\n")

# ───────────────────────────────────────────────────────────
# 2. EXERCISE DB — drop messy columns, clean what remains
# ───────────────────────────────────────────────────────────
print("2️⃣  Cleaning Exercise DB...")
df = pd.read_csv(f'{raw}/exercise_db/megaGymDataset.csv')
df = df.drop(columns=['Unnamed: 0', 'Desc', 'Rating', 'RatingDesc'])
df = df.dropna(subset=['Title', 'BodyPart'])  # need at least name + body part
df['Equipment'] = df['Equipment'].fillna('Bodyweight')
df.to_csv(f'{clean}/exercises_clean.csv', index=False)
print(f"   ✅ Kept {df.shape[0]} usable exercises. Final: {df.shape}\n")

# ───────────────────────────────────────────────────────────
# 3. GYM MEMBERS — already perfect, just standardize column names
# ───────────────────────────────────────────────────────────
print("3️⃣  Cleaning Gym Members...")
df = pd.read_csv(f'{raw}/gym_members/gym_members_exercise_tracking.csv')
df.columns = (df.columns.str.strip()
                          .str.replace(' ', '_')
                          .str.replace('(', '', regex=False)
                          .str.replace(')', '', regex=False)
                          .str.replace('/', '_per_'))
df.to_csv(f'{clean}/gym_members_clean.csv', index=False)
print(f"   ✅ Standardized columns. Final: {df.shape}\n")

# ───────────────────────────────────────────────────────────
# 4. NUTRITION — convert text-numbers to real numbers
# ───────────────────────────────────────────────────────────
print("4️⃣  Cleaning Nutrition...")
df = pd.read_csv(f'{raw}/nutrition/nutrients_csvfile.csv')

# 't' means trace amount — convert to 0
df = df.replace('t', 0)
df = df.replace('t\'', 0)
df = df.replace(',', '', regex=True)  # remove commas in numbers like "1,234"

# Convert all numeric columns
numeric_cols = ['Grams', 'Calories', 'Protein', 'Fat', 'Sat.Fat', 'Fiber', 'Carbs']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df = df.dropna(subset=['Calories', 'Protein'])  # need at least these
df.columns = df.columns.str.replace('.', '_', regex=False)
df.to_csv(f'{clean}/nutrition_clean.csv', index=False)
print(f"   ✅ Converted to numeric. Final: {df.shape}\n")

# ───────────────────────────────────────────────────────────
# 5. SLEEP HEALTH — fill missing sleep disorder as "None"
# ───────────────────────────────────────────────────────────
print("5️⃣  Cleaning Sleep Health...")
df = pd.read_csv(f'{raw}/sleep_health/Sleep_health_and_lifestyle_dataset.csv')
df['Sleep Disorder'] = df['Sleep Disorder'].fillna('None')

# Split blood pressure into systolic + diastolic
df[['Systolic_BP', 'Diastolic_BP']] = df['Blood Pressure'].str.split('/', expand=True).astype(int)
df = df.drop(columns=['Blood Pressure'])

df.columns = df.columns.str.strip().str.replace(' ', '_')
df.to_csv(f'{clean}/sleep_health_clean.csv', index=False)
print(f"   ✅ Filled missing, split blood pressure. Final: {df.shape}\n")

print("=" * 60)
print("🎉 ALL DATASETS CLEANED & SAVED!")
print("=" * 60)
print(f"\n📂 Clean files saved to:\n   {clean}\n")
````

## 5.3 Validating the Replacement DatasetBefore adopting the new dataset, we validate it against known nutritional reference values for common foods. A trustworthy dataset should return values close to published nutritional standards.

````python
import pandas as pd
import re

raw = '/content/drive/MyDrive/HELIX_Project/data/raw'
df = pd.read_csv(f'{raw}/nutrition_better/nutrition.csv')

# Drop the useless index column
df = df.drop(columns=['Unnamed: 0'])

print("🧪 SANITY CHECK — Real-world protein values\n")
print("="*70)

# Helper: clean up strings like "0.98 g" → 0.98
def clean_number(val):
    if pd.isna(val):
        return None
    if isinstance(val, (int, float)):
        return float(val)
    # Extract first number from string
    match = re.search(r'(\d+\.?\d*)', str(val))
    return float(match.group(1)) if match else None

# Test foods we know the real protein values for
test_foods = {
    'chicken breast': 31,   # Expected: ~31g per 100g
    'beef':           26,   # Expected: ~26g per 100g
    'oats':           17,   # Expected: ~17g per 100g
    'rice white':     2.7,  # Expected: ~2.7g per 100g
    'oysters':        9,    # Expected: ~9g per 100g
    'butter':         0.9,  # Expected: ~0.9g per 100g
    'beet':           1.6,  # Expected: ~1.6g per 100g
    'salmon':         20,   # Expected: ~20g per 100g
    'egg':            13,   # Expected: ~13g per 100g
}

print(f"{'Food':<30}{'Actual':<15}{'Expected':<15}{'Status'}")
print("-"*70)

for food, expected in test_foods.items():
    matches = df[df['name'].str.contains(food, case=False, na=False)].head(1)
    if len(matches) > 0:
        protein_val = clean_number(matches['protein'].iloc[0])
        if protein_val is not None:
            # Check if within 60% of expected (reasonable tolerance)
            diff_pct = abs(protein_val - expected) / expected * 100
            status = "✅ CORRECT" if diff_pct < 60 else "❌ WRONG"
            food_name = matches['name'].iloc[0][:28]
            print(f"{food_name:<30}{protein_val:<15.2f}{expected:<15}{status}")
        else:
            print(f"{food:<30}Could not parse")
    else:
        print(f"{food:<30}Not found")

print("\n" + "="*70)
print("💡 If most values are CORRECT, this dataset is trustworthy!")
print("="*70)
````

## 5.4 Final Cleaning & ReplacementWith the new dataset validated, we apply final cleaning (parse string values like "12.5 g" into numeric values) and save it as the canonical `nutrition_clean.csv`, replacing the broken version.

````python
import pandas as pd
import re
import os

raw = '/content/drive/MyDrive/HELIX_Project/data/raw'
clean = '/content/drive/MyDrive/HELIX_Project/data/processed'

df = pd.read_csv(f'{raw}/nutrition_better/nutrition.csv')
df = df.drop(columns=['Unnamed: 0'])

print("🔬 FINAL SANITY CHECK — using precise search terms\n")
print("="*70)

# Helper to extract numbers from strings like "12.5 g" or "1,200 mg"
def clean_number(val):
    if pd.isna(val):
        return None
    if isinstance(val, (int, float)):
        return float(val)
    val = str(val).replace(',', '')
    match = re.search(r'(\d+\.?\d*)', val)
    return float(match.group(1)) if match else None

# Clean nutrient columns (these all have "5.2 g" type strings)
nutrient_cols = ['calories', 'protein', 'carbohydrate', 'fiber', 'fat',
                 'saturated_fat', 'leucine', 'sugars']

for col in nutrient_cols:
    if col in df.columns:
        df[col] = df[col].apply(clean_number)

# Verify with precise food searches
verifications = [
    ('Chicken raw breast', 'chicken, broilers or fryers, breast, meat only, raw', 23),
    ('Beef raw',           'beef, ground, 80% lean meat',                          18),
    ('Rice white cooked',  'rice, white, long-grain, regular, cooked',             2.7),
    ('Eggs whole raw',     'egg, whole, raw, fresh',                               13),
    ('Salmon raw',         'fish, salmon, atlantic, farmed, raw',                  20),
    ('Whey-like protein',  'whey, sweet, dried',                                   12),
]

print(f"{'Test Food':<22}{'Found':<45}{'Protein':<10}{'OK?'}")
print("-"*85)

for label, search, expected in verifications:
    matches = df[df['name'].str.lower().str.contains(search.lower(), na=False, regex=False)].head(1)
    if len(matches) > 0:
        protein = matches['protein'].iloc[0]
        food_name = matches['name'].iloc[0][:43]
        diff = abs(protein - expected) / expected * 100 if expected > 0 else 0
        status = "✅" if diff < 50 else "❌"
        print(f"{label:<22}{food_name:<45}{protein:<10.2f}{status}")
    else:
        print(f"{label:<22}Not found")

# Top high-protein foods (real check)
print(f"\n\n🏆 TOP 10 REAL HIGH-PROTEIN FOODS:")
print("-"*70)
top10 = df.nlargest(10, 'protein')[['name', 'calories', 'protein']]
print(top10.to_string(index=False))

# Save cleaned version
df.to_csv(f'{clean}/nutrition_clean.csv', index=False)
print(f"\n\n✅ Clean nutrition data saved: {df.shape}")
print(f"📂 Saved to: {clean}/nutrition_clean.csv")

# Remove old broken file
old_path = f'{raw}/nutrition/nutrients_csvfile.csv'
if os.path.exists(old_path):
    print(f"\n🗑️  Old broken file kept in raw/ for transparency")
    print(f"   New clean file: nutrition_clean.csv now uses the better dataset")
````
