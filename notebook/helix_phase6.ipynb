```python
import os

# First: is Drive even mounted?
if not os.path.exists('/content/drive/MyDrive'):
    print("⚠️ Drive not mounted — mounting now...")
    from google.colab import drive
    drive.mount('/content/drive')
else:
    print("✅ Drive is mounted\n")

ROOT = '/content/drive/MyDrive/HELIX_Project'

# Show the top-level project structure
print("Contents of HELIX_Project/:")
for f in sorted(os.listdir(ROOT)):
    full = os.path.join(ROOT, f)
    tag = "📁" if os.path.isdir(full) else "  "
    print(f"  {tag} {f}")

# Now hunt for model files anywhere under the project
print("\nSearching for model artifacts (.pt / .joblib / .ttl / .owl):")
for dirpath, dirnames, filenames in os.walk(ROOT):
    for fn in filenames:
        if fn.endswith(('.pt', '.joblib', '.ttl', '.owl')):
            p = os.path.join(dirpath, fn)
            kb = os.path.getsize(p) / 1024
            print(f"  {kb:8.1f} KB   {p}")
```
